# Experiment 25: VMD Hand-Crafted Features
**Configuration**:
- Active Branches = `[0]` (VMD Features)
- Wavelet Type = `vmd_feats`
- CSV Suffix = `'_exp25'`

**Features Extracted per IMF (5 IMFs total)**:
1. **Energy**
2. **Statistical**: Mean, Variance, Std Dev, Kurtosis, Skewness
3. **Entropy**: Shannon Entropy
4. **Frequency-domain**: Spectral Centroid, Bandwidth, Peak Frequency

Total features: 5 IMFs × 10 features = 50 features.

```python
import sys; sys.path.append('../')
from src.dataset import get_dataloaders
from src.train_eval import run_experiment
from src.models.LEEDNet_mini import LEEDNet_mini
import torch

# 1. Hyperparameters
active_branches = [0]
wavelet_type = 'vmd_feats'
csv_suffix = '_exp25' 
epochs = 50

# 2. Data Loading
train_loader, val_loader, test_loader = get_dataloaders(
    csv_suffix=csv_suffix, 
    active_branches=active_branches, 
    wavelet_type=wavelet_type
)

# 3. Model Initialization (Input feature sizes: [1, 33, 30])
# Branch 0 receives the 50 VMD features
model = LEEDNet_mini(input_channels_list=[1, 33, 30], output_size=3)

# 4. Training and Evaluation Pipeline
results = run_experiment(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    test_loader=test_loader,
    active_branches=active_branches,
    model_save_path=f'../dataset/model/weights/exp25.pt',
    pred_save_path=f'../dataset/predictions/exp25.csv',
    epochs=epochs
)
```

```python
# 5. Standalone Evaluation
from src.train_eval import evaluate_on_test_set

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
evaluate_on_test_set(
    model=model,
    test_loader=test_loader,
    active_branches=active_branches,
    model_save_path='../dataset/model/weights/exp25.pt',
    pred_save_path='../dataset/predictions/exp25_eval.csv',
    device=device
)
```
