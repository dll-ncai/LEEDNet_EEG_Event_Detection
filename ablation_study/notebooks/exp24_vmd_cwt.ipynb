# Experiment 24: Preprocessed VMD -> CWT Pipeline
**Configuration**:
- Active Branches = `[0, 2]`
- Wavelet Type = `vmd_cwt` (Sequential processing)
- CSV Suffix = `'_exp24'` (Precomputed sequential features)

**Input Pipeline (Precomputed)**:
1. Signal decomposed into 5 VMD IMFs.
2. CWT applied to each IMF.
3. Averaged 5 CWT maps into a single `[30, 400]` feature stored in the `exp24` preprocessed folder.
4. Experiment fused with Branch 0 (Raw EEG).

```python
import sys; sys.path.append('../')
from src.dataset import get_dataloaders
from src.train_eval import run_experiment
from src.models.LEEDNet_mini import LEEDNet_mini
import torch

# 1. Hyperparameters
active_branches = [0, 2]
wavelet_type = 'vmd_cwt'
csv_suffix = '_exp24' # Using precomputed features
epochs = 50

# 2. Data Loading
train_loader, val_loader, test_loader = get_dataloaders(
    csv_suffix=csv_suffix, 
    active_branches=active_branches, 
    wavelet_type=wavelet_type
)

# 3. Model Initialization (Input feature sizes: [1, 33, 30])
model = LEEDNet_mini(input_channels_list=[1, 33, 30], output_size=3)

# 4. Training and Evaluation Pipeline
results = run_experiment(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    test_loader=test_loader,
    active_branches=active_branches,
    model_save_path=f'../dataset/model/weights/exp24.pt',
    pred_save_path=f'../dataset/predictions/exp24.csv',
    epochs=epochs
)
```

```python
# 5. Standalone Evaluation
from src.train_eval import evaluate_on_test_set

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
evaluate_on_test_set(
    model=model,
    test_loader=test_loader,
    active_branches=active_branches,
    model_save_path='../dataset/model/weights/exp24.pt',
    pred_save_path='../dataset/predictions/exp24_eval.csv',
    device=device
)
```
